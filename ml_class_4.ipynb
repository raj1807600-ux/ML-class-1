```python
from sklearn.preprocessing import PowerTransformer
import numpy as np
```

```python
X=np.array([
    [-1],
    [0],
    [3],
    [5],
    [10],
    [50]
])
#create the transformer
pt = PowerTransformer(method='yeo-johnson')
#fit and transform
X_transformed = pt.fit_transform(X)



print('original')
print(X.ravel())

print('\nTransformed:')
print(np.round(X_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
y=np.array([
    [1],
    [2],
    [3],
    [5],
    [10],
    [50]
])
#create the transformer
pt = PowerTransformer(method='box-cox')
#fit and transform
y_transformed = pt.fit_transform(y)



print('original')
print(y.ravel())

print('\nTransformed:')
print(np.round(y_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
import pandas as pd
import numpy as np

# Dataset with Missing Values
data = {
    'Emp_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Name': ['Amit', 'Priya', 'Rahul', 'Sneha', 'Vikas', 'Pooja', 'Ankit', 'Neha', 'Rohan', 'Kavita'],
    
    # --- Numerical Columns (Containing Null Values) ---
    'Age': [25, np.nan, 29, 32, np.nan, 27, 35, np.nan, 28, 31],
    'Salary': [50000, 62000, np.nan, 75000, 48000, np.nan, 90000, 58000, np.nan, 67000],
    'Experience_Years': [2.0, 5.5, 4.0, np.nan, 1.5, 3.0, np.nan, 4.5, 3.5, np.nan],
    
    # --- Categorical Columns (Containing Null Values) ---
    'Department': ['IT', 'HR', np.nan, 'Finance', 'IT', 'Marketing', np.nan, 'HR', 'IT', 'Finance'],
    'City': ['Delhi', 'Mumbai', 'Bangalore', np.nan, 'Delhi', np.nan, 'Pune', 'Mumbai', 'Delhi', np.nan],
    'Performance': ['Good', 'Excellent', np.nan, 'Average', 'Good', 'Excellent', 'Average', np.nan, 'Good', 'Excellent']
}

df = pd.DataFrame(data)
df
```

```python
df.head()
```

```python
# 1. Null values count check karein
print("\n--- Missing Values Count ---")
print(df.isnull().sum())

# 2. Null values percentage (%) check karein
print("\n--- Missing Values Percentage ---")
print((df.isnull().sum() / len(df)) * 100)
```
