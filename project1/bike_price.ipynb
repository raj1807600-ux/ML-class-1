```python
from IPython.terminal.shortcuts.auto_match import braces
def get_number(value):
    if pd.isna(value,):
        return np.nan
    value=str(value).replace(',','')
    number='' #string me value store hai
    dot_used=False

    for i in value:
        if i.isdigit():
            number+=i
        elif i==',' and dot_used:
            dot_used = True
        elif number:
            break
        return float(number ) if number else np.nan
```
